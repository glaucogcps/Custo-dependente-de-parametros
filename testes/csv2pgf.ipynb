```python
import matplotlib
matplotlib.use('pgf')
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib import cm

plt.rcParams.update({
    'pgf.texsystem': 'pdflatex',
    'font.family': 'serif',
    'font.size': 8,
    'text.usetex': True,
    'pgf.rcfonts': False,
    'figure.figsize': (3.5, 2.6),
    'pgf.preamble': r"""
\usepackage[utf8]{inputenc}
\usepackage[T1]{fontenc}
\usepackage{amsmath}
\usepackage{amssymb}
\usepackage{mathtools}
\usepackage{dsfont}
"""
})

df = pd.read_csv('teste_controle_Hinf_MJLS_Ex7_N2_indep_PontosGrafico.csv')
graus = sorted(df['Grau_rho'].unique())

fig, ax = plt.subplots()

colors = ['r', 'b', 'g', 'c', 'm', 'k']

for i, grau in enumerate(graus):
    subdf = df[df['Grau_rho'] == grau].sort_values('Alpha')
    ax.plot(subdf['Alpha'], subdf['Custo_Real'], color=colors[i], linestyle='-', linewidth=1.5, label=rf'$d = {grau}$')
    ax.plot(subdf['Alpha'], subdf['Custo_Garantido'], color=colors[i], linestyle='--', linewidth=1.5, label=rf'$d = {grau}$')

ax.set_xlabel(r'$\alpha$')
ax.set_ylabel(r'$\mathcal{{H}}_\infty$')
# ax.set_ylabel(r'$\mathcal{{H}}_2$')
ax.legend(loc='best')
ax.grid(True, linestyle=':', alpha=0.5)

plt.tight_layout()
plt.savefig('teste_controle_Hinf_MJLS_Ex7_N2_indep.pgf', bbox_inches='tight')
plt.close()
```

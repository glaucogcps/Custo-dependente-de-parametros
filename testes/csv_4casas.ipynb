```python
import pandas as pd
import numpy as np

# Assuming 'nome_arquivo' is defined, e.g.,
nome_arquivo = 'sistema_teste_d_Hinf_Tabela_Hp.csv'

# Load the CSV
df = pd.read_csv(nome_arquivo)

# Identify float columns
float_cols = df.select_dtypes(include=[np.float64, np.float32]).columns.tolist()

# Format float columns to scientific notation with 4 decimal places using lambda and map
df[float_cols] = df[float_cols].map(lambda x: f'{x:.4e}' if pd.notna(x) else '')

# Save with suffix '_edited.csv'
nome_arquivo_edited = nome_arquivo.rsplit('.', 1)[0] + '_edited.csv'
df.to_csv(nome_arquivo_edited, index=False)

print(f'File saved as: {nome_arquivo_edited}')
```

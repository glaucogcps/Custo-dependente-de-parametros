```python
import pandas as pd
import os

# O usuário só precisa alterar o valor da variável 'nome_arquivo' abaixo para o nome do seu arquivo CSV.
nome_arquivo = 'teste_controle_Hinf_MJLS_Ex7_N2_indep_Tabela.csv'

# Leitura do arquivo CSV
df = pd.read_csv(nome_arquivo)

# Arredonda todos os valores float para 4 casas decimais
numeric_columns = df.select_dtypes(include=['float64', 'float32']).columns
df[numeric_columns] = df[numeric_columns].round(4)

# Gera o nome do arquivo de saída com sufixo '_edited'
base_name = os.path.splitext(nome_arquivo)[0]
output_file = f'{base_name}_edited.csv'

# Salva o DataFrame editado
df.to_csv(output_file, index=False)

print(f'Arquivo editado salvo como: {output_file}')
```

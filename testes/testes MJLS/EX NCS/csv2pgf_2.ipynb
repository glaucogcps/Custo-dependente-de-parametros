```python
import matplotlib
matplotlib.use('pgf')
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Configurações de backend e preâmbulo LaTeX
plt.rcParams.update({
    'pgf.texsystem': 'pdflatex',
    'font.family': 'serif',
    'font.size': 8,
    'text.usetex': True,
    'pgf.rcfonts': False,
    'figure.figsize': (3.5, 2.6), # Dimensão exata da coluna IEEE
    'pgf.preamble': r"""
\usepackage[utf8]{inputenc}
\usepackage[T1]{fontenc}
\usepackage{amsmath}
\usepackage{amssymb}
\usepackage{mathtools}
\usepackage{dsfont}
"""
})

# Carregamento dos dados
df = pd.read_csv('teste_controle_Hinf_MJLS_Ex7_N2_indep_PontosGrafico.csv')
graus = sorted(df['Grau_rho'].unique())

# GERANDO O GRÁFICO PRINCIPAL
fig2, ax2 = plt.subplots()
colors = ['r', 'b', 'g', 'c', 'm', 'k']

for i, grau in enumerate(graus):
    subdf = df[df['Grau_rho'] == grau].sort_values('Alpha')
    cor_atual = colors[i % len(colors)]
    
    # Plotamos e guardamos os handles para a legenda posterior
    ax2.plot(subdf['Alpha'], subdf['Custo_Real'], color=cor_atual, 
            linestyle='-', linewidth=1.5, label=rf'$d = {grau}$')
    ax2.plot(subdf['Alpha'], subdf['Custo_Garantido'], color=cor_atual, 
            linestyle='--', linewidth=1.5,)

ax2.set_xlabel(r'$\alpha$')
ax2.set_ylabel(r'$\mathcal{H}_\infty$')
ax2.grid(True, linestyle=':', alpha=0.5)

# Salva o gráfico sem a legenda
plt.savefig('teste_controle_Hinf_MJLS_Ex7_N2_indep.pgf', bbox_inches='tight')

# GERANDO A LEGENDA SEPARADA
# Extraímos os handles e labels do gráfico original
handles, labels = ax2.get_legend_handles_labels()

# Criamos uma nova figura apenas para a legenda
fig_leg = plt.figure(figsize=(2, 1)) # Tamanho inicial arbitrário
ax_leg = fig_leg.add_subplot(111)

# Adicionamos a legenda centralizada nesta nova figura
legend = ax_leg.legend(handles, labels, loc='center', ncol=1, fontsize=7, frameon=True)

# Removemos os eixos, molduras e ticks da figura da legenda
ax_leg.axis('off')

# Salva apenas a legenda, cortando todo o espaço em branco excedente
fig_leg.savefig('legenda_MJLS.pgf', bbox_inches='tight')

plt.close('all')
```

```python
import matplotlib
matplotlib.use('pgf')
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.ticker import MultipleLocator

plt.rcParams.update({
    'pgf.texsystem': 'pdflatex',
    'font.family': 'serif',
    'font.size': 8,
    'text.usetex': True,
    'pgf.rcfonts': False,
    'figure.figsize': (3.5, 2.6),
    'pgf.preamble': r"""
\usepackage[utf8]{inputenc}
\usepackage[T1]{fontenc}
\usepackage{amsmath}
\usepackage{amssymb}
\usepackage{mathtools}
\usepackage{dsfont}
"""
})

df = pd.read_csv('teste_controle_H2_MJLS_NCS_N2_PontosGrafico.csv')
graus = sorted(df['Grau_rho'].unique())

fig3, ax3 = plt.subplots()
colors = ['r', 'b', 'g', 'c', 'm', 'k']

# Definição dos limites reais baseados no seu código MATLAB
pmin = 0.1
pmax = 0.4

for i, grau in enumerate(graus):
    subdf = df[df['Grau_rho'] == grau].sort_values('Alpha')
    cor_atual = colors[i % len(colors)]
    
    # REPRODUZINDO A ESCALA DO MATLAB: pmax * alpha + pmin * (1 - alpha)
    # Isso transforma o Alpha (0 a 1) para o intervalo (0.1 a 0.4)
    x_real = pmax * subdf['Alpha'] + pmin * (1 - subdf['Alpha'])
    
    ax3.plot(x_real, subdf['Custo_Real'], color=cor_atual, linestyle='-', label=rf'$d={grau}$')
    ax3.plot(x_real, subdf['Custo_Garantido'], color=cor_atual, linestyle='--')

ax3.set_xlabel(r'Failure probability')
ax3.set_ylabel(r'$\mathcal{H}_2$')

# Forçamos os limites do eixo X para o intervalo real do problema
ax3.set_xlim(pmin, pmax)
ax3.set_ylim(2, 6)

# --- AJUSTE DE ESCALA E GRID ---
# Eixo X: marcas de 0.2 em 0.2 (ex: 0.2, 0.4)
ax3.xaxis.set_major_locator(MultipleLocator(0.05))
# Eixo Y: marcas de 5 em 5 (ex: 10, 15, 20, 25)
ax3.yaxis.set_major_locator(MultipleLocator(0.5))

# Grid reforçado (alpha=1.0 e cor silver conforme solicitado)
ax3.grid(True, which='major', linestyle=':', alpha=1.0, color='silver')

# Legenda interna
ax3.legend(loc='lower right', frameon=True, fancybox=True, framealpha=0.8, fontsize=6)

plt.savefig('teste_controle_H2_MJLS_NCS_N2.pgf', bbox_inches='tight')
plt.close()
```

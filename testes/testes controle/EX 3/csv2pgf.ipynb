```python
import matplotlib
matplotlib.use('pgf')
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib import cm

plt.rcParams.update({
    'pgf.texsystem': 'pdflatex',
    'font.family': 'serif',
    'font.size': 8,
    'text.usetex': True,
    'pgf.rcfonts': False,
    'figure.figsize': (3.5, 2.6),
    'pgf.preamble': r"""
\usepackage[utf8]{inputenc}
\usepackage[T1]{fontenc}
\usepackage{amsmath}
\usepackage{amssymb}
\usepackage{mathtools}
\usepackage{dsfont}
"""
})

df = pd.read_csv('teste_controle_H2_Ex3_N2_PontosGrafico.csv')
graus = sorted(df['Grau_rho'].unique())

fig1, ax1 = plt.subplots()

colors = ['r', 'b', 'g', 'c', 'm', 'k']

for i, grau in enumerate(graus):
    subdf = df[df['Grau_rho'] == grau].sort_values('Alpha')
    # Seleciona a cor do vetor
    cor_atual = colors[i % len(colors)]
    
    ax1.plot(subdf['Alpha'], subdf['Custo_Real'], color=cor_atual, linestyle='-', linewidth=1.5)
    ax1.plot(subdf['Alpha'], subdf['Custo_Garantido'], color=cor_atual, linestyle='--', linewidth=1.5)

ax1.set_xlabel(r'$\alpha$')
ax1.set_ylabel(r'$\mathcal{H}_2$')
ax1.grid(True, linestyle=':', alpha=0.5)

# O bbox_inches='tight' é crucial aqui para que a legenda externa não seja cortada
plt.savefig('teste_controle_H2_Ex3_N2.pgf', bbox_inches='tight')
plt.close()
```

```python
import matplotlib
matplotlib.use('pgf')
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Configurações de backend e preâmbulo LaTeX
plt.rcParams.update({
    'pgf.texsystem': 'pdflatex',
    'font.family': 'serif',
    'font.size': 8,
    'text.usetex': True,
    'pgf.rcfonts': False,
    'figure.figsize': (3.5, 2.6), # Dimensão exata da coluna IEEE
    'pgf.preamble': r"""
\usepackage[utf8]{inputenc}
\usepackage[T1]{fontenc}
\usepackage{amsmath}
\usepackage{amssymb}
\usepackage{mathtools}
\usepackage{dsfont}
"""
})

# Carregamento dos dados
df = pd.read_csv('teste_controle_Hinf_Ex3_N2_PontosGrafico.csv')
graus = sorted(df['Grau_rho'].unique())

# --- GERANDO O GRÁFICO PRINCIPAL ---
fig2, ax2 = plt.subplots()
colors = ['r', 'b', 'g', 'c', 'm', 'k']

for i, grau in enumerate(graus):
    subdf = df[df['Grau_rho'] == grau].sort_values('Alpha')
    cor_atual = colors[i % len(colors)]
    
    # Plotamos e guardamos os handles para a legenda posterior
    ax2.plot(subdf['Alpha'], subdf['Custo_Real'], color=cor_atual, 
            linestyle='-', linewidth=1.5)
    ax2.plot(subdf['Alpha'], subdf['Custo_Garantido'], color=cor_atual, 
            linestyle='--', linewidth=1.5,)

ax2.set_xlabel(r'$\alpha$')
ax2.set_ylabel(r'$\mathcal{H}_\infty$')
ax2.grid(True, linestyle=':', alpha=0.5)

# Salva o gráfico sem a legenda
plt.savefig('teste_controle_Hinf_Ex3_N2_indep.pgf', bbox_inches='tight')

# --- GERANDO A LEGENDA SEPARADA ---
# Extraímos os handles e labels do gráfico original
handles, labels = ax2.get_legend_handles_labels()

# Criamos uma nova figura apenas para a legenda
fig_leg = plt.figure(figsize=(2, 1)) # Tamanho inicial arbitrário
ax_leg = fig_leg.add_subplot(111)

# Adicionamos a legenda centralizada nesta nova figura
legend = ax_leg.legend(handles, labels, loc='center', ncol=6, fontsize=7, frameon=True)

# Removemos os eixos, molduras e ticks da figura da legenda
ax_leg.axis('off')

# Salva apenas a legenda, cortando todo o espaço em branco excedente
fig_leg.savefig('legenda_LTI.pgf', bbox_inches='tight')

plt.close('all')
```
